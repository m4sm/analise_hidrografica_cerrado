```python
import pandas as pd

# 1. Carregar a planilha das estações
df = pd.read_excel('Estacoes Cerrado.xlsx')

# 2. Corrigir as escalas de dados (se estiverem com fator 1e15)
colunas_ajuste = [
    'MED_ANUAL',
    'MED_CDD',
    'MED_JAN',
    'MED_FEV',
    'MED_MAR',
    'MED_ABR',
    'MED_MAI',
    'MED_JUN',
    'MED_JUL',
    'MED_AGO',
    'MED_SET',
    'MED_OUT',
    'MED_NOV',
    'MED_DEZ',
]
for col in colunas_ajuste:
  if col in df.columns and df[col].max() > 10000:
    df[col] = df[col] / 1e15

# 3. Selecionar colunas essenciais para o dashboard
colunas_finais = [
    'Nome',
    'Latitude',
    'Longitude',
    'Altitude',
    'Municipio',
    'Estado',
    'MED_ANUAL',
    'MED_CDD',
    'MED_JAN',
    'MED_FEV',
    'MED_MAR',
    'MED_ABR',
    'MED_MAI',
    'MED_JUN',
    'MED_JUL',
    'MED_AGO',
    'MED_SET',
    'MED_OUT',
    'MED_NOV',
    'MED_DEZ',
]

df_export = df[[c for c in colunas_finais if c in df.columns]]
df_export.to_csv('estacoes_cerrado_dashboard.csv', index=False, encoding='utf-8')
print('CSV gerado com sucesso!')
```
